```python
Certificate_fixed = []

for movie in Movie_Name:
    certificate = movie.find_elements(
        "xpath",
        "./ancestor::li[1]//div[contains(@class,'cli-title-metadata')]//li[3]"
    )

    if len(certificate) > 0:
        Certificate_fixed.append(certificate[0].text)
    else:
        Certificate_fixed.append(None)
```
