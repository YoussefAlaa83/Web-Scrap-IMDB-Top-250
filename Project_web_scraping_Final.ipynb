```python
movies = []

for i in range(len(Movie_Name)):
    movies.append({
        "Movie_Name": Movie_Name[i].text,
        "Release_Year": Release_Year[i].text,
        "Runtime": Runtime[i].text,
        "Certificate": Certificate_fixed[i],
        "Rate": Rate[i].text
    })

df = pd.DataFrame(movies)
```

```python
df['Movie_Name'] = df['Movie_Name'].str.strip()
df['Certificate'] = df['Certificate'].str.strip()
```

```python
df['Release_Year'] = df['Release_Year'].astype(int)
```

```python
df['Certificate'] = df['Certificate'].fillna('Not Rated')
```

```python
df
```

```python
df['Rate'] = df['Rate'].astype(float)
```

```python
df.drop_duplicates(inplace=True)
df.reset_index(drop=True,inplace=True)
```

```python
import matplotlib.pyplot as plt
rates = []
durations = []
for movie in movies:
    runtime = movie["Runtime"]
    try:
        if "h" in runtime:
            parts = runtime.split("h")
            hours = int(parts[0])
            minutes = 0
            if len(parts) > 1:
                minutes_text = parts[1].replace("m", "").strip()
                if minutes_text:
                    minutes = int(minutes_text)
            total_minutes = hours * 60 + minutes
            durations.append(total_minutes)
            rates.append(float(movie["Rate"]))
    except:
        continue
plt.figure(figsize=(10,6))
plt.scatter(durations, rates)
plt.xlabel("Duration (minutes)")
plt.ylabel("Rating")
plt.title("Movie Duration vs Rating")
plt.grid(True)
plt.show()
```

```python
year_counts = df['Release_Year'].value_counts().sort_index()

plt.figure(figsize=(12,6))
year_counts.plot(kind='bar', color='skyblue')
plt.title("Number of Movies per Release Year")
plt.xlabel("Release Year")
plt.ylabel("Number of Movies")
plt.grid(axis='y')
plt.show()
```

```python
top_certs = df['Certificate'].value_counts()
main_certs = top_certs.iloc[:4]
main_certs['Other'] = top_certs.iloc[4:].sum()

plt.figure(figsize=(7, 7))
plt.pie(main_certs, labels=main_certs.index, autopct='%1.1f%%')
plt.title('Movie Distribution by Certificate')
plt.show()
```
